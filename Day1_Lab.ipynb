```python
voltage = 3
current = 1.5 
resistance = voltage / current
print("Resistance:", resistance, "Ohms")
```

```python
power = voltage * current
print("Power:", power, "Watts")
```

```python
resistors = [100, 220, 330, 470, 500, 680, 1000]
print("Resistor values (Ohms):", resistors)
total_resistance = sum(resistors)
print("Total Resistance:", total_resistance, "Ohms")
print ("First resistor value:", resistors[0], "Ohms")
```
