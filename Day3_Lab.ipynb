```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
time_sec = np.linspace(0, 50, 100)
voltage = 4.2 - 0.1 * time_sec - 0.01 * time_sec**2
plt.figure(figsize=(8, 5))
plt.plot(time_sec, voltage)
plt.xlabel("Time (seconds)")
plt.ylabel("Voltage (V)")
plt.title("Li-ion Battery Discharge Curve")
plt.grid(True)
plt.legend(["Voltage"])
plt.show()
```

```python
time_sec=np.linspace(0, 10, 500)
current = 10 * np.sin(time_sec)
noise = np.random.normal(0, 1, 500)
sensor_reading = current + noise
plt.figure(figsize=(8, 5))
plt.plot(time_sec, sensor_reading, label="Sensor Reading", color='blue', alpha=0.6)
plt.plot(time_sec, current, label="True Current", color='orange', linewidth=3)
plt.xlabel("Time (seconds)")
plt.ylabel("Current (A)")
plt.title("Current Sensor Readings with Noise")
plt.grid(True)
plt.legend(["True Current", "Sensor Reading"])
plt.show()

indices = np.where(sensor_reading > 11)[0]
count = indices.size
print('Indices (0-based):', indices)
print('Number of readings > 11 A:', count)
```
